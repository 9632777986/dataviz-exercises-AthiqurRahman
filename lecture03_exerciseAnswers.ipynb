```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)

df = pd.read_csv('/content/co2_emissions.csv')

print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python

# Keep only Asian countries
asia_df = df[df['Region'] == 'Asia'].copy()

# Choose China as the highlighted country
highlight_country = 'China'

fig = go.Figure()

# Plot the other Asian countries in light grey
for country in asia_df['Country'].unique():
    if country == highlight_country:
        continue

    country_df = (
        asia_df[asia_df['Country'] == country]
        .sort_values('Year')
    )

    fig.add_trace(go.Scatter(
        x=country_df['Year'],
        y=country_df['CO2_Mt'],
        mode='lines',
        line=dict(color='#EEEEEE', width=1),
        hovertemplate=(
            f'{country}<br>'
            'Year: %{x}<br>'
            'CO2 emissions: %{y:.1f} Mt'
            '<extra></extra>'
        ),
        showlegend=False
    ))

# Highlight China
china_df = (
    asia_df[asia_df['Country'] == highlight_country]
    .sort_values('Year')
)

fig.add_trace(go.Scatter(
    x=china_df['Year'],
    y=china_df['CO2_Mt'],
    mode='lines+markers+text',
    line=dict(color='#E76F51', width=3.5),
    marker=dict(size=4, color='#E76F51'),
    text=[''] * max(0, len(china_df) - 1) + ['China'],
    textposition='middle right',
    textfont=dict(size=12, color='#E76F51'),
    hovertemplate=(
        'China<br>'
        'Year: %{x}<br>'
        'CO2 emissions: %{y:.1f} Mt'
        '<extra></extra>'
    ),
    showlegend=False
))

# Format the chart
fig.update_layout(
    title=dict(
        text="China's CO2 emissions rose sharply across 2000–2022",
        x=0.02,
        xanchor='left'
    ),
    xaxis_title='Year',
    yaxis_title='CO2 emissions (Mt)',
    template='plotly_white',
    plot_bgcolor='white',
    paper_bgcolor='white',
    showlegend=False,
    margin=dict(l=60, r=100, t=80, b=50),
    hovermode='closest'
)

# Keep the chart clean and readable
fig.update_xaxes(
    showgrid=False,
    showline=False,
    zeroline=False,
    dtick=5
)

fig.update_yaxes(
    showgrid=True,
    gridcolor='#E6E6E6',
    showline=False,
    zeroline=False
)

fig.show()

```
